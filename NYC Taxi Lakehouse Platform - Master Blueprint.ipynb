### Executive Summary

The client requires an **end-to-end Data Engineering Lakehouse platform** built on the NYC Taxi public dataset. The platform must ingest raw taxi trip data, process it through a multi-layer Medallion Architecture (Bronze → Silver → Gold), enforce data quality, and deliver business analytics dashboards — all using **free tools and publicly available data**.

The solution will be built on **Databricks** using **PySpark, Spark SQL, and Delta Lake**. No paid subscriptions are required for data sources, storage, or visualization tools.

---

### Business Context

New York City's Taxi & Limousine Commission (TLC) publishes millions of trip records every month — pickup times, dropoff times, distances, fares, and locations. The client wants to turn this raw data into **actionable business insights**: revenue trends, peak demand hours, zone-level analysis, and operational metrics.

The platform must be production-ready, scalable, well-monitored, and demonstrate industry best practices including incremental loading, data quality validation, and performance optimization.

## Architecture Overview

### Landing Zone
The project begins by ingesting raw NYC Taxi datasets into a **Landing Zone** where files are preserved exactly as received.

### Bronze Layer
A **Bronze Layer** is created using Delta tables to store raw data along with ingestion metadata such as:
- Load timestamps
- Source file information

### Silver Layer
The **Silver Layer** contains cleansed, validated, and transformed datasets. Data engineering operations include:
- Deduplication
- Standardization
- Schema enforcement
- Type casting
- Derived column generation

The data will be partitioned appropriately to optimize storage and query performance.

### Gold Layer
The **Gold Layer** serves as the business and analytics layer. Aggregated datasets will be created to answer business questions related to:
- Revenue trends
- Trip patterns
- Demand forecasting
- Peak travel periods
- Customer behavior
- Geographic analysis
- Operational efficiency

## Data Loading Strategies

The platform will support both **Full Load** and **Incremental Load** strategies:

| Strategy | Description |
| --- | --- |
| **Full Load** | Used during the initial ingestion of all historical data |
| **Incremental Load** | Subsequent executions use watermark-based incremental loading with a metadata table tracking previously processed records |

This approach reflects industry-standard ETL and ELT design patterns to reduce processing costs and improve scalability.

## Data Quality Framework

A reusable **Data Quality Framework** validates incoming data before promotion to downstream layers:

| Check | Description |
| --- | --- |
| Null validation | Ensure critical columns are not null |
| Duplicate detection | Identify and remove duplicate records |
| Datatype verification | Confirm column types match expected schema |

| Negative value checks | Flag invalid negative values in numeric fields |
| Record count reconciliation | Cross-check row counts between layers |
| Audit logging | Persist all quality results to audit tables |

Results are stored in quality audit tables for observability and reporting.

# Phase 2: Landing & Bronze Layer — Detailed Documentation

## What This Phase Does

The Landing & Bronze layer is the **first data touchpoint** in the Medallion Architecture. Raw data from the source is placed into a landing zone (files as-is), then loaded into a Bronze Delta table with ingestion metadata appended.

---

## Step-by-Step Breakdown

### Step 1: Create the Landing Zone

* **What:** A Unity Catalog Volume (`nyc_taxi.landing.raw_files`) acts as a file-based staging area
* **Why:** Keeps raw files untouched before any transformation — supports reprocessing and auditability
* **How:** Raw Parquet/CSV files from the NYC TLC source are written into the volume path `/Volumes/nyc_taxi/landing/raw_files/`

### Step 2: Load Raw Data into Landing Zone

* **What:** Read the source dataset (e.g., `samples.nyctaxi.trips`) and write it as Parquet files into the landing volume
* **Why:** Parquet is columnar, compressed, and optimized for Spark reads — far better than CSV for large datasets
* **Format choice:** Parquet (columnar, compressed) vs CSV (row-based, human-readable)

### Step 3: Create Bronze Delta Table

* **What:** Read the raw Parquet files from the landing zone and save them as a managed Delta table (`nyc_taxi.bronze.trips_raw`)
* **Ingestion metadata columns added:**
  * `_load_timestamp` — when this batch was loaded into Bronze
  * `_source_file` — the source file path (using `_metadata.file_path`)
  * `_ingest_date` — derived from pickup datetime, used as the partition column
* **Partitioning:** By `_ingest_date` so queries filtering by date only scan relevant partitions

---

## Key Concepts to Understand

| Concept | Explanation |
| --- | --- |
| **Delta Lake** | Storage layer that adds ACID transactions, schema enforcement, and time travel to Parquet files |
| **Managed Table** | Databricks manages both the metadata and the underlying data files |
| **Partitioning** | Physically organizes data into folders by a column value — speeds up filtered queries |
| **_metadata.file_path** | Built-in Spark column that returns the source file name for each row — replaces `input_file_name()` which is not supported in Unity Catalog |
| **Landing Zone** | Immutable raw file area — data is never modified here, only read |

# Phase 3: Incremental Load Framework — Detailed Documentation

## What This Phase Does

Instead of reloading the entire dataset every time, the incremental load framework **only processes new or changed data** since the last run. This dramatically reduces processing time and cost.

---

## Problem with Full Load

| Issue | Impact |
| --- | --- |
| Reloads all data every run | Wastes compute and storage |
| No tracking of what was processed | Cannot guarantee no duplicates or missed rows |
| Not scalable for large datasets | Monthly data = millions of rows — full reload is impractical |

---

## Solution: Watermark-Based Incremental Load

### What is a Watermark?
A watermark is a **marker value** (usually a timestamp) that records the last point up to which data has been processed. On the next run, only data **after** the watermark is loaded.

### How It Works

```
Last run watermark:  2016-01-15 23:59:59

This run:
  → Read all rows WHERE tpep_pickup_datetime > '2016-01-15 23:59:59'
  → Process only those new rows
  → Update watermark to MAX(tpep_pickup_datetime) in this batch
  → Next run starts from the new watermark
```

---

## Components of the Framework

### 1. Metadata Control Table (`nyc_taxi.bronze.load_metadata`)

Tracks every load run with key information:

| Column | Description |
| --- | --- |
| `load_id` | Unique ID for each load run |
| `load_timestamp` | When the load ran |
| `source_table` | Where data came from |
| `target_table` | Where data was written |
| `watermark_value` | Last processed timestamp |
| `rows_processed` | Number of rows loaded in this batch |
| `status` | SUCCESS / FAILED |
| `load_type` | FULL / INCREMENTAL |

### 2. Watermark Logic

* **First run:** No watermark exists → perform a **full load**
* **Subsequent runs:** Read the watermark from the metadata table → load only rows after the watermark → append to Bronze table
* **Watermark update:** After a successful load, update the metadata table with the new maximum timestamp

### 3. Append Mode

* Incremental loads use `mode("append")` — new rows are added without touching existing data
* The Bronze table grows over time as new batches arrive

---

## Full Load vs Incremental Load

| Aspect | Full Load | Incremental Load |
| --- | --- | --- |
| When to use | First run, or full backfill | Every subsequent scheduled run |
| Data volume processed | Entire dataset | Only new rows since last watermark |
| Mode | Overwrite | Append |
| Watermark needed | No | Yes |
| Speed | Slow (full scan) | Fast (filtered scan) |

---

## Key Concepts to Understand

| Concept | Explanation |
| --- | --- |
| **Watermark** | A stored timestamp marking the boundary between processed and unprocessed data |
| **Metadata table** | A control table that tracks every pipeline run — its status, row counts, and watermark |
| **Idempotency** | If you re-run the pipeline, it should not create duplicates — the watermark prevents reprocessing |
| **Append mode** | Adds new rows to the existing Delta table without overwriting |
| **MAX() function** | Used to find the latest timestamp in the current batch to update the watermark |

# Phase 4: Data Quality Framework — Detailed Documentation

## What This Phase Does

Before data moves from Bronze to Silver, it must pass a series of **quality checks**. This phase builds a reusable framework that validates data, logs results, and prevents bad data from reaching downstream layers.

---

## Why Data Quality Matters

| Without DQ | With DQ Framework |
| --- | --- |
| Bad data silently flows into analytics | Bad data is caught and flagged before it propagates |
| No visibility into data health | Audit tables show pass/fail for every check |
| Downstream reports are wrong | Only clean data reaches Gold layer and dashboards |
| Hard to debug issues | Quality logs point directly to the problem |

---

## Quality Checks Implemented

### 1. Null Validation
* **What:** Ensure critical columns are not NULL
* **Columns checked:** `tpep_pickup_datetime`, `tpep_dropoff_datetime`, `trip_distance`, `fare_amount`
* **Action:** Rows with nulls in critical columns are flagged and excluded from Silver

### 2. Duplicate Detection
* **What:** Identify rows that appear more than once
* **Method:** Group by all business columns, count occurrences, flag rows with count > 1
* **Action:** Keep only one copy (first occurrence), remove duplicates

### 3. Negative Value Checks
* **What:** Flag rows where numeric columns have invalid negative values
* **Columns checked:** `trip_distance`, `fare_amount`
* **Action:** Negative values are set to NULL or the row is excluded

### 4. Business Rule Validation
* **What:** Domain-specific rules that make sense for taxi data
* **Rules:**
  * Dropoff datetime must be >= pickup datetime (trip cannot end before it starts)
  * Trip distance must be > 0 (a trip must cover some distance)
  * Fare amount must be >= 0 (minimum fare is $0)

### 5. Record Count Reconciliation
* **What:** Compare row counts between Bronze and Silver layers
* **Why:** Ensures no data was lost or duplicated during transformation
* **Action:** Log counts in an audit table; alert if counts don't match expectations

---

## Quality Audit Table (`nyc_taxi.bronze.quality_audit`)

Every check result is persisted to an audit table for observability:

| Column | Description |

# Phase 7: Delta Lake Features — Detailed Documentation

## What This Phase Does

This phase demonstrates the advanced capabilities of Delta Lake that make it more powerful than plain Parquet or CSV. These features are what make a Lakehouse a Lakehouse.

---

## Delta Lake Features Demonstrated

### 1. ACID Transactions
* **What:** Multiple reads and writes can happen concurrently without data corruption
* **Why it matters:** In a real pipeline, multiple jobs may write to the same table — ACID guarantees no dirty reads or lost updates
* **How:** Delta uses a transaction log (`_delta_log/`) to track every change

### 2. Time Travel
* **What:** Query previous versions of a table as of a specific version number or timestamp
* **Syntax:**
  * `SELECT * FROM nyc_taxi.bronze.trips_raw VERSION AS OF 5`
  * `SELECT * FROM nyc_taxi.bronze.trips_raw TIMESTAMP AS OF '2026-09-18 10:00:00'`
* **Use cases:**
  * Audit what data looked like before a change
  * Rollback to a previous version if a bad load happened
  * Compare data between two points in time

### 3. MERGE (Upsert)
* **What:** Insert new rows AND update existing rows in a single operation
* **When to use:** Incremental loads where some rows are new and some are updates to existing records
* **How it works:**
  * Match on a key column (e.g., pickup_datetime + pickup_zip)
  * If match found → UPDATE the existing row
  * If no match → INSERT the new row
* **Why better than INSERT + UPDATE separately:** Atomic, single-pass, no race conditions

### 4. Schema Enforcement & Evolution
* **Schema enforcement:** Delta rejects writes that don't match the table schema — prevents bad data from corrupting the table
* **Schema evolution:** You can add new columns to a Delta table without rewriting existing data
  * `ALTER TABLE nyc_taxi.silver.trips_clean ADD COLUMN (new_column STRING)`
* **Merge schema:** `df.write.option("mergeSchema", "true").saveAsTable(...)` automatically adds new columns

### 5. OPTIMIZE (File Compaction)
* **What:** Combines many small files into fewer larger files
* **Why:** Too many small files slow down reads — Spark has to open each file separately
* **Command:** `OPTIMIZE nyc_taxi.bronze.trips_raw`
* **ZORDER BY:** `OPTIMIZE nyc_taxi.bronze.trips_raw ZORDER BY (pickup_zip)`
  * Z-ordering physically co-locates rows with similar values — dramatically speeds up queries filtering by that column

### 6. VACUUM (Storage Cleanup)
* **What:** Removes old data files no longer referenced by the Delta transaction log
* **Why:** Time travel keeps old files around — VACUVM reclaims that storage space
* **Command:** `VACUUM nyc_taxi.bronze.trips_raw RETAIN 168 HOURS`
  * Keeps files newer than 7 days (168 hours); removes older ones
* **Caution:** After VACUUM, you cannot time travel to versions older than the retention period

### 7. Transaction History Inspection
* **What:** View the history of all operations performed on a Delta table
* **Command:** `DESCRIBE HISTORY nyc_taxi.bronze.trips_raw`
* **Shows:** Each operation (write, merge, optimize), timestamp, user, version number
* **Use case:** Audit trail — see exactly what happened and when

---

## Delta Lake vs Traditional Storage

| Feature | CSV | Parquet | Delta Lake |
| --- | --- | --- | --- |
| ACID transactions | No | No | Yes |
| Time travel | No | No | Yes |
| MERGE / Upsert | No | No | Yes |
| Schema enforcement | No | No | Yes |
| Schema evolution | No | No | Yes |
| File compaction (OPTIMIZE) | No | No | Yes |
| Storage cleanup (VACUUM) | No | No | Yes |
| Transaction log / audit | No | No | Yes |

---

* How Delta Lake's transaction log provides ACID guarantees over file storage
* How Time Travel enables auditing and rollback without maintaining separate copies

* How OPTIMIZE + ZORDER improve read performance by co-locating data

# Phase 8: Performance Optimization — Detailed Documentation

## What This Phase Does

This phase applies Spark and Delta Lake optimization techniques to make queries faster and more efficient. Understanding these is critical for handling large datasets.

---

## Optimization Techniques

### 1. Partitioning
* **What:** Physically splits data into directories by a column value
* **Why:** Spark only scans the partitions that match the filter — skips the rest
* **Best practice:** Partition by columns frequently used in WHERE clauses (e.g., date)
* **Caution:** Over-partitioning (too many small partitions) creates many tiny files — slows down reads
* **Applied in this project:** Bronze partitioned by `_ingest_date`, Silver by `trip_date`

### 2. Caching
* **What:** Stores a DataFrame in memory so repeated operations don't recompute from disk
* **When to use:** When the same DataFrame is queried multiple times in a notebook
* **Command:** `df.cache()` followed by an action like `df.count()` to materialize
* **When NOT to use:** If the DataFrame is only used once — caching adds overhead with no benefit

### 3. Broadcast Joins
* **What:** When joining a large table with a small table, Spark broadcasts the small table to all executors
* **Why:** Eliminates the shuffle (network data transfer) phase — massive speedup
* **Threshold:** Tables smaller than 10MB are auto-broadcast by default (`spark.sql.autoBroadcastJoinThreshold`)
* **Manual control:** `df.hint("broadcast")` or `broadcast(small_df)`
* **Example in this project:** Joining trip data (millions of rows) with a small zone lookup table

### 4. Adaptive Query Execution (AQE)
* **What:** Spark dynamically adjusts the query plan at runtime based on actual data statistics
* **Features:**
  * Coalesces small partitions into larger ones after shuffle
  * Switches from sort-merge join to broadcast join if one side turns out to be small
  * Splits skewed partitions into smaller ones (skew handling)
* **Enabled by default** in Databricks (`spark.sql.adaptive.enabled = true`)

### 5. File Compaction (OPTIMIZE)
* **What:** Combines many small files into fewer larger files
* **Why:** Reduces the number of files Spark needs to open — improves read performance
* **Command:** `OPTIMIZE nyc_taxi.silver.trips_clean`
* **ZORDER BY:** Co-locates rows with similar values for a column — speeds up point queries
* **Command:** `OPTIMIZE nyc_taxi.silver.trips_clean ZORDER BY (pickup_zip)`

### 6. Query Plan Analysis (EXPLAIN)
* **What:** `df.explain()` shows the physical execution plan Spark will use
* **Why:** Helps identify bottlenecks — shuffles, scans, joins
* **What to look for:**
  * `Exchange` → indicates a shuffle (network data movement — can be slow)
  * `BroadcastHashJoin` → good (no shuffle)
  * `SortMergeJoin` → may be slow for large tables
  * `Filter` pushed down to the source → good (early filtering)

### 7. Repartitioning
* **What:** Manually control the number of partitions in a DataFrame
* **When to use:**
  * `repartition(n)` — increase partitions for parallelism (good before a large shuffle)
  * `coalesce(n)` — decrease partitions without a shuffle (good before writing files)
* **Caution:** Don't over-repartition — too many partitions means too many small tasks

---

## Performance Troubleshooting Checklist

| Symptom | Likely Cause | Solution |
| --- | --- | --- |
| Query is very slow | Full table scan | Add partition filter in WHERE clause |
| Too many small files | Over-partitioning | Run OPTIMIZE to compact files |
| Join is slow | Sort-merge join on large tables | Use broadcast join if one side is small |
| Out-of-memory errors | Partition skew | Enable AQE or repartition by a different key |
| Repeated queries slow | No caching | Cache the DataFrame if used multiple times |
